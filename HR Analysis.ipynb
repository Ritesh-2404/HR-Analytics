```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("HR_Data.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["YearsWithCurrManager"] = df["YearsWithCurrManager"].fillna(
    df["YearsWithCurrManager"].median()
)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.drop(
    columns=[
        "EmployeeCount",
        "Over18",
        "StandardHours"
    ],
    inplace=True
)
```

```python
df.dtypes
```

```python
for col in df.columns:
    print(col)
    print(df[col].unique())
    print("-"*50)
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18,25,35,45,55,65],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65"
    ]
)
```

```python
df["IncomeBand"] = pd.cut(
    df["MonthlyIncome"],
    bins=[0,5000,10000,15000,20000,50000],
    labels=[
        "0-5K",
        "5K-10K",
        "10K-15K",
        "15K-20K",
        "20K+"
    ]
)
```

```python
df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=[0,2,5,10,20,40],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)
```
